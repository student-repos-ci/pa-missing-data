```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pingouin as pg
```

```python
df = pd.read_csv('diabetes_prediction_dataset.csv')
df.head(3)
```

```python
pg.normality(data=df, alpha=0.05)
```

```python
variable = 'blood_glucose_level'
```

```python
print(df['gender'].value_counts())
av_var_by_gender = df.groupby(by=('gender')).agg(bgl=(variable, 'mean'))
av_var_by_gender
```

```python
df_filtered = df[df['gender'].isin(['Male', 'Female'])]
df_filtered['gender'].value_counts()
```

```python
male_var = df[variable][df['gender'] == 'Male']
male_var = male_var.sample(n=1000, random_state=111)
```

```python
female_var = df[variable][df['gender'] == 'Female']
female_var = female_var.sample(n=1000, random_state=111)
```

```python
pg.mwu(x=male_var, y=female_var)
```
