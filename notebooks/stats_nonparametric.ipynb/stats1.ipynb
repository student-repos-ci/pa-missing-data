```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pingouin as pg
```

The data from Statistical Tests Topic 3: Non-parametric tests

```python
df = pd.DataFrame(data={'Col3':[18.3, 13.3, 16.5, 12.6, 9.5, 13.6, 8.1, 8.9, 10, 8.3, 7.9, 8.1, 13.4],
                        "Col4":[12.7, 11.1, 15.3, 12.7, 10.5, 15.6, 11.2, 14.2, 16.3, 15.5, 19.9, 20.4, 36.1]
                        })
df.head()
```

#### Plot Col3 and Col4 separately for visual clarity

```python
sns.histplot(data=df, bins=10, x='Col3', kde=True)
plt.show()
```

Calculate normality for Col3

```python
pg.normality(data=df['Col3'], alpha=0.05)
```

And skewness for Col3

```python
skew_col3 = df['Col3'].skew()
skew_col3
```

#### Plot Col4 separately

```python
sns.histplot(data=df, x='Col4', kde=True)
plt.show()
```

```python
pg.normality(data=df['Col4'], alpha=0.05)
```

```python
skew_col4 = df['Col4'].skew()
skew_col4
```

#### Plot a qq plot for both columns

The red lines represent the confidence interval boundaries. If the data is normally distributed, 95% of the data should fall within these boundaries

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))

pg.qqplot(df['Col3'], dist='norm', ax=axes[0])
pg.qqplot(df['Col4'], dist='norm', ax=axes[1])
plt.show()
```
